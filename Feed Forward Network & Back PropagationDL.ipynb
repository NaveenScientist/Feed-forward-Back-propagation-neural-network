# Feed Forward Neural Network:

```python
from sklearn.neural_network import MLPClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

# Load the iris dataset

```python
data = load_iris()

X = data.data
y = data.target
```

```python
# view the data:

data
```

```python
# split the data into training and test sets:

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.2, random_state = 52)
```

```python
# defining the neural network model:
# hidden_layer_size = (64,32): two hidden layers with 64 and 32 neurons:
model = MLPClassifier(hidden_layer_sizes=(64,32), activation = "relu" , max_iter = 300, random_state= 52)
```

```python
# train the model

model.fit(X_train, y_train)
```

```python
# assign the weights and biases

coefs = model.coefs_
intercepts = model.intercepts_
```

```python
# priniting he weights and biases for eah layer

for i, (weights,biases) in enumerate(zip(coefs, intercepts)):
    print(f"weights shape: {weights.shape}")
    print(f"biases shape: {biases.shape}")
    print(f"weights: {weights}")
    print(f"biases: {biases}\n")
```

```python
# predict on the test set
y_pred = model.predict(X_test)
```

```python
#calculate accuracy:

accuracy = accuracy_score(y_test,y_pred)
print(f"accuracy: {accuracy}")
```

```python
# Activation Functions Visualization
import numpy as np
import matplotlib.pyplot as plt

# Define activation functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

# Derivatives
def dsigmoid(x):
    s = sigmoid(x)
    return s * (1 - s)

def dtanh(x):
    return 1 - np.tanh(x) ** 2

def drelu(x):
    return np.where(x > 0, 1, 0)

# Generate data
x = np.linspace(-5, 5, 200)

# Plot each activation and derivative
plt.figure(figsize=(14, 10))

# Sigmoid
plt.subplot(3, 2, 1)
plt.plot(x, sigmoid(x), label='Sigmoid')
plt.title("Sigmoid Function")
plt.grid(True)
plt.legend()

plt.subplot(3, 2, 2)
plt.plot(x, dsigmoid(x), label='Sigmoid Derivative', color='red')
plt.title("Sigmoid Derivative")
plt.grid(True)
plt.legend()

# Tanh
plt.subplot(3, 2, 3)
plt.plot(x, tanh(x), label='Tanh')
plt.title("Tanh Function")
plt.grid(True)
plt.legend()

plt.subplot(3, 2, 4)
plt.plot(x, dtanh(x), label='Tanh Derivative', color='red')
plt.title("Tanh Derivative")
plt.grid(True)
plt.legend()

# ReLU
plt.subplot(3, 2, 5)
plt.plot(x, relu(x), label='ReLU')
plt.title("ReLU Function")
plt.grid(True)
plt.legend()

plt.subplot(3, 2, 6)
plt.plot(x, drelu(x), label='ReLU Derivative', color='red')
plt.title("ReLU Derivative")
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# 📘 Simple Neural Network (1 neuron) Forward and Backward Pass

# Input (x), Weight (w), Bias (b), and Target (y)
x = 0.5
y_true = 1
w = 0.8
b = 0.2
lr = 0.1  # learning rate

# Choose activation
def forward_sigmoid(x, w, b):
    z = w * x + b
    a = sigmoid(z)
    return z, a

def forward_relu(x, w, b):
    z = w * x + b
    a = relu(z)
    return z, a

# Forward pass (Sigmoid)
z, a = forward_sigmoid(x, w, b)
loss = 0.5 * (y_true - a)**2

# Backward pass
dz = (a - y_true) * dsigmoid(z)
dw = dz * x
db = dz

# Update weights
w -= lr * dw
b -= lr * db

print("=== Sigmoid Neuron ===")
print(f"Output (a): {a:.4f}")
print(f"Loss: {loss:.6f}")
print(f"Updated Weight: {w:.4f}, Bias: {b:.4f}")
```
